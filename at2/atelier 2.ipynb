# Partie 2 - Manipulation de tableux a plusieurs dimensions

```python
Temperatures = np.array([[22, 25, 20, 19, 24, 28, 30], 
                         [18, 21, 19, 22, 25, 27, 26], 
                         [30, 32, 29, 28, 27, 33, 35], 
                         [15, 17, 16, 18, 19, 20, 21], 
                         [25, 26, 24, 27, 29, 31, 30]
                        ])
```

```python
Precipitations = np.random.uniform(0, 20, (5, 7))
Precipitations.round(1)
```

```python
print(Temperatures.shape)
print(Precipitations.shape)
```

#### Quelle est la différence entre np.array() et np.random.uniform() ?
###### np.array() crée un tableau à partir de valeurs données, tandis que np.random.uniform() génère des valeurs aléatoires entre deux limites.

```python
jours = np.arange(1, 8)
print(jours)
```

```python
t_transpose = Temperatures.T
print(t_transpose)
```

```python
produit = np.dot(Temperatures, Precipitations.T)
print(produit)
```
